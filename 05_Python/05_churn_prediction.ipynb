# Churn Prediction Prototype
Logistic Regression using engagement, completion, session, support, NPS and value indicators.


```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, classification_report
ROOT=Path.cwd()
if not (ROOT/'02_Data').exists(): ROOT=ROOT.parent
df=pd.read_csv(ROOT/'02_Data/processed/customer_360.csv')
features=['engagement_score','course_completion_rate','avg_session_duration_minutes','support_ticket_count','nps_score','total_sessions','courses_enrolled','total_revenue']
X=df[features].fillna(0); y=df['churn_flag']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
model=LogisticRegression(max_iter=1000); model.fit(X_train,y_train)
proba=model.predict_proba(X_test)[:,1]
print('ROC-AUC:',roc_auc_score(y_test,proba))
print(classification_report(y_test,model.predict(X_test)))
```

```python
pd.DataFrame({'feature':features,'coefficient':model.coef_[0]}).sort_values('coefficient',ascending=False)
```
