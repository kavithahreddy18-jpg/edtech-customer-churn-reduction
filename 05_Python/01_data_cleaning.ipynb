# EdTech Churn — Data Cleaning & EDA
This notebook loads the processed customer-level analytical dataset, validates fields, calculates KPIs and explores churn by segment.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'02_Data').exists(): ROOT=ROOT.parent
df=pd.read_csv(ROOT/'02_Data/processed/customer_360.csv')
df.head()
```

```python
kpis=pd.Series({'Total Customers':len(df),'Churn Rate':df.churn_flag.mean(),'Retention Rate':1-df.churn_flag.mean(),'Completion Rate':df.course_completion_rate.mean(),'Avg Session Minutes':df.avg_session_duration_minutes.mean(),'ARPU':df.arpu.mean(),'Estimated LTV':df.estimated_ltv.mean()})
kpis
```

```python
segment=df.groupby('customer_segment').agg(customers=('customer_id','count'),churn_rate=('churn_flag','mean'),engagement=('engagement_score','mean'),completion=('course_completion_rate','mean')).sort_values('churn_rate',ascending=False)
segment
```

```python
segment['churn_rate'].plot(kind='bar',title='Churn Rate by Segment',ylabel='Churn Rate')
plt.tight_layout(); plt.show()
```
